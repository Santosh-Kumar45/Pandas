#Importing pandas

```python
import numpy as np
import pandas as pd
```

#Series from list

```python
#string

country=['Indai','pakistan','nepal','dubai','sri lanka']
pd.Series(country)
```

```python
#integer

runs=[53,23,55,93,24]
pd.Series(runs)
```

```python
#custom series
subject=['english','science','sst','hindi','maths']
marks=[93,34,83,20,92]
pd.Series(marks,index=subject)
```

```python
#setting name
pd.Series(marks,index=subject,name='santosh ke marks')
```

```python
#using Dictionaries

marks={
    "english":78,
    "hindi":44,
    "maths":47
}

marks_series=pd.Series(marks ,name="santosh ke marks")
marks_series
```

# Series Attributes

```python
#size
marks_series.size
```

```python
#dtype
marks_series.dtype
```

```python
#name
marks_series.name
```

```python
#is_unique
marks_series.is_unique
```

```python
#values
marks_series.values
```
