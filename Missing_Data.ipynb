## Finding missing Data

```python
import pandas as pd
import numpy as np

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [3, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().any()
```

```python
#isna() it check the nan -true, false
df.isna()
```

## Removing missing data

```python
df
```

```python
df.dropna()
```

```python
#no of thresh =no of non-null values in row so,that will be printing 
df.dropna(thresh=3)
```

## Filling the missing data

```python
df
```

```python
#temporary change
df.fillna(0)
```

```python
#value assign by using columns name
value={
    'A':0, 
    'B':100, 
    'C':200, 
    'D':300
}
df.fillna(value=value)
```
